## 演算子とオペランド

JupyterLab Desktop で利用しているnotebook 形式の入力は，電卓のように利用できますので使ってみましょう。`7 + 3` を入力して，`Shift + Enter`を入力するか，左側の(‣)マークを押すと計算が実行され，その計算結果が下に表示されます。

```python
7 + 3
```

```python
7 + 3
7 - 3
```

複数行の計算を行なったとき，計算は1行ごとで行なわれます。表示される結果は一番最後の行の計算結果が表示されます。（notebook形式の場合。）

次はかけ算と割り算です。

```python
7 * 3
```

```python
7 / 3
```

`*` は数字のかけ算を行なう演算記号(演算子という)です。割り算は３種類あり，`/`は実数の割り算，`//`は余りを切り捨てた整数値，`%`は割り算の余りを表示します。

```python
7 // 3
```

```python
7 % 3
```


べき乗の演算は`**`で表します。MS エクセルの関数では`^`を使ったりしますが，Pythonでは`**`を使いますので気をつけてください。

```python
2 ** 3
```

`^`はべき乗とは異なる計算を表しています。エラーが表示されませんので気をつけてください。（難しい演算です。）

```python
2 ^ 3
```

今出てきた`+` `-` `*` `/` などを演算子といいます。演算子の前や後にあるものをオペランドと言います。演算子による計算には優先順位があることに気をつけてください。

```python
7 - 3 / 2
```

```python
(7 - 3) / 2
```

左にある演算子から計算されるように思いますが，`/`は`-`より優先されます。このように左から計算したい場合は`()`を付けましょう。括弧は計算の順番が優先されます。

数値（データ）を表す方法（扱い方）は，プログラミング言語によって異なります。表す方法の種類は型(type)と呼ばれています。

```python
type(7)
```

```python
type(3)
```

```python
type(7 + 3)
```

画面への表示を行なう関数は`print()`です。

１行目の `#` は脚注を開始する文字です。プログラムの内容を説明するためにコメントを入れて，他の開発者や数年後忘れた自分に宛てて説明を行なうものです。`#`の左側にくる文字列はブログラムの動作に影しません。2行目の空行もプログラムに影響しません。プログラムを読みやすいように空行を入れましょう。

```python
print('こんにちは。\nはじめまして。')
```

`\n`を文字列の中に入れると改行されます。

```python
# プログラムの実行の順番を確認する。

print('1')
print('2')
print('3')
```

```python
print('好きな果物は？')
fruit = input()
print(fruit,'は，おいしいですね。')
```

`input()`関数はキーボードから入力した文字列を読み込みます。`fruit`変数に代入しました。`print()`関数に`,`をいれることにより複数の語句をいれることができます。`,`の位置に空白（スペース）が入ることを確認しておきましょう。

```python
print(fruit, 'は新潟の発酵食品です')
```

```python
type(fruit)
```

```python
food = input('好きなおつまみは？：')
print(food, 'は日本酒に合いますね。')
```

`input()`の引数（`()`の中に入れるもの）に文字列を入れると，文字列が表示された後で入力が促されます。`print()`の引数で与えられた文字列や変数は`,`で区切られていてスペースが代わりに入って表示されます。

```python
print(food, 'は日本酒に合いますね。')
print('1','2','3','引数の間にはスペースが入ります。')
```

スペースの代わりに何も入れたくないときは`sep=''`を引数に指定しましょう。


```python
print(food, 'は日本酒に合いますね。',sep='')
print('1','2','3',sep='')
```

プログラムの中身を見るとs1とs2を足して，その結果を表示させていることがわかりますが，3+5は8ですので，表示された結果はおかしいですよね？何が起きているのでしょうか？`type()`で確認してみましょう。

s1もs2もs1+s2も型がstr つまり文字列であることがわかります。じつは数字を入れていても，数値として認識されず，数字を表現する文字として認識されていることがわかります。ですので数字を入力と書いてあっても，文字を入力してもプログラムは実行されます。

```python
s1 = input('好きな数字を入力してください：')
s2 = input('好きな数字を入力してください：')
print('type(s1)=',type(s1),'type(s2)=',type(s2))
print(s1 + s2)
print('type(s1+s2)=',type(s1+s2))
```

文字列を数値に変換するには`int()`関数や`float()`を用います。

```python
s1 = input('好きな数字を入力してください：')
s2 = input('好きな数字を入力してください：')
print('type(s1)=',type(s1),'type(s2)=',type(s2))
summ = int(s1) + int(s2)
print(s1, '+', s2, '=', summ)
print('type(summ)=',type(summ))
```

数値に変換されたs1,s2は四則演算が可能です。その結果正しい（望んでいた）結果が表示されました。

```python
int('abc')
```

数字以外の文字列はint()により数値に変換することはできません。

## 文字列とリスト

文字列は実は後で説明するリストと同じような操作が可能です。

例えば文字列の中の文字一つ一つを指定して表示するなどが可能です。

文字列を`a`に代入したとき，一つ目の文字は`a[0]`，二つ目の文字は`a[1]`などと指定できます。Pythonでリストの位置を指定するときは0から数え始めます。覚えておきましょう。

```python
a = "ろいはにほへと"
print(a[0])
print(a[1])
print(a[2])
print(a[3])
print(a[4])
print(a[5])
print(a[6])
```

文字列の文字の個数は関数`len()`を使うと分かります。

```python
len(a)
```

位置を指定する数字にマイナスを付けると文字列の語尾から数えられます。

```python
print(a[-1])
print(a[-2])
print(a[-3])
```
